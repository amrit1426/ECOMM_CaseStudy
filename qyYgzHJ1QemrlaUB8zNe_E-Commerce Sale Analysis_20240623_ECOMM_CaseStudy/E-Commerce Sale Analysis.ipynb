<a name = Section62></a>
### **6.2 Handling of Missing Data**

- In this section we will see if we can **fill** the missing data with some **logical imputation.**

```python
missing_cat_cols = ['Gender', 'City_Category', 'Product_Category']
missing_num_cols = []  # [], (), {}, pd.Series(), arrays   -> always Iterable even when empty
```

```python
for col in missing_cat_cols:
    mode_value = df[col].mode()[0]
    df[col] = df[col].fillna(mode_value)
```

```python
for col in missing_num_cols:
    median_value = df[col].median()
    df[col] = df[col].fillna(median_value)
```

```python
# Cross-Checking the Operations
df.isnull().sum()
```

**Observation:**

- Missing Values in the **Gender and City_category** has been **Filled Successfully.**

- Now, to fill the missing values in **Product_Category**:

    - We have **Product ID** in the dataset, so we can use that information to fill the missing values in **Product_Category.**

    - First create a **Dictionary** that **maps** every **Product_ID** to it\'s **corresponding Product_Category.**

    - Use that **Dictionary to fill the missing Product_Category** if that product was ordered in any of the rest of the purchases.

    - Finally **drop any rows** in which **we can not fill Product_Category** using the above described method.

**Observation:**

- All the **Missing values in the Product Category has been filled** using the information of ProductID

#### PARETO ANALYSIS  (80-20 Principle)

```python
product_sales = (df
                .groupby('ProductID')
                .agg({'Purchase': 'sum'})
                .sort_values('Purchase', ascending=False))

# A. How many products contribute to 80% Sales!
proportion_target = 0.8

product_sales_perc = (product_sales / product_sales.sum())
product_sales_perc = product_sales_perc.cumsum()

products_contribution_sales = (product_sales_perc
                               .query("Purchase <= @proportion_target")
                               .shape[0])

print(f"{products_contribution_sales} products contribute to {proportion_target*100}% of sales")

### B. Can I show this in a visualisation
```

```python
pareto_df = product_sales_perc.reset_index()
pareto_df.index = pareto_df.index + 1
pareto_df['Purchase'] = pareto_df['Purchase'] * 100
```

```python
fig, ax = plt.subplots(figsize=(14, 5))

pareto_df['Purchase'].plot(kind='line', ax=ax)

plt.axvline(products_contribution_sales, linestyle=':', color='r')
plt.axhline(80, linestyle=':', color='r')
plt.annotate(products_contribution_sales, xy=(products_contribution_sales, 5))

plt.grid(linestyle=':')
plt.ylabel("Percentage of Sales")
plt.xlabel("Count of Products")
plt.title("Product Sales Pareto Analysis")

plt.tight_layout()
```
